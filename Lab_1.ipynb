```python
import numpy as np
import pandas as pd
print("Numpy version: ",np.__version__)
print("Pandas version: ",pd.__version__)
```

```python
marks = np.array([43,45,78,89,76])
print("Type :",type(marks))
```

```python
print("Add 5:",marks+5)
print("Multiply by 2:", marks*2)
print("Square:",marks**2)
```

```python
print("Sum:",marks.sum())
print("Mean:",marks.mean())
print("Variance:",marks.var())
print("Maximum:",marks.max())
print("Minimum:",marks.min())
```

```python
print("First:",marks[0])
print("last:",marks[-1])
```

```python
print("Index 1 to 3:",marks[1:4])
print("First 3:",marks[:3])
print("Every second element:",marks[::2])
```

```python
s_marks = np.array([
    [88,92,72],
    [62,84,79],
    [46,79,85],
])
print(s_marks)
print("Shape:",s_marks.shape)
```

```python
print("Mean of all values:", s_marks.mean())
print("Column wise mean:", s_marks.mean(axis=0))
print("Row-wise mean:",s_marks.mean(axis=1))
```
