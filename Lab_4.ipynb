```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
data = {
    "Customer_ID": ["C101","C102","C103","C104","C105","C106","C107","C108","C109","C110"],
    "City": ["Bengaluru","Mysuru","Hubballi","Bengaluru","Mangaluru",
             "Mysuru","Hubballi","Bengaluru","Mangaluru","Mysuru"],
    "Age": [25, np.nan, 31, 28, 42, 35, np.nan, 29, 38, 27],
    "Income": [45000, 38000, np.nan, 52000, 47000, 41000, 36000, np.nan, 55000, 40000],
    "Membership": ["Gold","Silver","Gold",np.nan,"Silver",
                   "Gold","Bronze","Gold","Silver","Bronze"],
    "Purchase": [3200,2100,4100,3500,2800,3900,1800,4600,5200,2400]
}

df = pd.DataFrame(data)
df
```

```python
print("Shape: ",df.shape)
print("Data type")
print(df.dtypes)
print("First 5 rows")
print(df.head())
print("Last 5 rows")
print(df.tail)
```

```python
missing_count = df.isnull().sum()
missing_count
```

```python
missing_percentage = (df.isnull().sum()/len(df))
missing_percentage
```

```python
age_mean = df["Age"].mean()
df_mean = df.copy()
df_mean["Age"].fillna(age_mean,inplace=True)
df_mean["Age"]
```

```python
income_mean = df['Income'].median()
df_median = df.copy()
df_median['Income'].fillna(income_mean,inplace=True)
df_median['Income']
```

```python
values = pd.Series([20,25,30,35,100])
print("Mean: ",values.mean())
print("Median: ",values.median())
```

```python
weather = pd.DataFrame({
    "Day": ["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday"],
    "Temperature": [30,np.nan,32,np.nan,31,33]
})
weather
```

```python
weather_ffill = weather.copy()
weather_ffill['Temperature'] = weather_ffill['Temperature'].ffill()
weather_ffill
```

```python
copydf = df.copy()
copydf["Age"] = copydf["Age"].fillna(age_mean)
copydf["Income"] = copydf["Income"].fillna(income_mean)
copydf.isnull().sum()
```

```python
membership_mode = copydf["Membership"].mode()[0]
copydf['Membership'] = copydf['Membership'].fillna(membership_mode)
print(membership_mode)
copydf['Membership']
```

```python
encoded_df = pd.get_dummies(copydf,columns=['City','Membership'],dtype=int)
encoded_df
```

```python
print("Missing values")
print(encoded_df.isnull().sum())
print("Data types")
print(encoded_df.dtypes)
print("Final Shape:",encoded_df.shape)
```
