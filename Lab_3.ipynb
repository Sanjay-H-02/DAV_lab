```python
plt.figure(figsize=(10,6))
plt.hist(df['Final_Score'],bins=8,edgecolor='black')
plt.title("Distribution of Final Scores")
plt.xlabel("Final Score")
plt.ylabel("Number of students")
plt.show()
```

```python
 plt.figure(figsize=(10,6))
 sns.boxplot(df['Final_Score'])
 plt.title("Boxplot of Final Scores")
 plt.ylabel("Final Score")
 plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df,x="Department",y="Final_Score",hue="Department")
plt.xlabel("Department")
plt.ylabel("Final Score")
plt.title("Final Score by Department")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Study_Hours"],df["Final_Score"])
plt.title("Study Hours vs Final Score")
plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x="Study_Hours",y="Final_Score",hue="Department",s=80)
plt.title("Study Hours vs Final Score")
plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.show()
```

```python
 num_cols = ['Attendance', 'Study_Hours', 'Assignment_Score', 'Internal_Marks', 'Final_Score']
 sns.pairplot(df[num_cols])
 plt.show()
```

```python
corr = df[num_cols].corr()
display(corr)
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(corr,annot=True,cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
